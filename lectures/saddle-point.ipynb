```python
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
import numpy as np

# plotting a saddle point in 3d
fig = plt.figure()
ax = plt.axes(projection='3d')
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 - Y**2
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.show()
```
